```python
import pandas as pd
```

#### Load raw data

```python
# Load data
train = pd.read_csv("../data/train.csv")
test = pd.read_csv("../data/test.csv")

print("Training:", train.shape)
print("Test:", test.shape)
```

# Data Preprocessing
Preparing the Airline Passenger Satisfaction dataset for classification.

```python
# Remove identifiers Unnamed: 0 and id
# we don't want the classification tree learning splits based on record identifiers.
train = train.drop(columns=["Unnamed: 0", "id"])
test = test.drop(columns=["Unnamed: 0", "id"])

print("Training:", train.shape)
print("Test:", test.shape)
```

```python
# Check missing values in the training and test datasets
print("Training missing values:")
print(train.isnull().sum()[train.isnull().sum() > 0])

print("\nTest missing values:")
print(test.isnull().sum()[test.isnull().sum() > 0])

print("\nTraining rows affected:", train.isnull().any(axis=1).sum())
print("Test rows affected:", test.isnull().any(axis=1).sum())
```

#### Missing Value Handling

```python
# Remove rows containing missing values; 
# the number of missing rows is very small, so removing them has 
# minimal effect on the dataset size

train = train.dropna().reset_index(drop=True)
test = test.dropna().reset_index(drop=True)

print("Training:", train.shape)
print("Test:", test.shape)
print("Remaining missing values:", train.isnull().sum().sum(), test.isnull().sum().sum())
```

#### Final Data Check

```python
# Check the final dataset sizes and confirm that no missing values remain

print("Training:", train.shape)
print("Test:", test.shape)

print("\nMissing values in training:", train.isnull().sum().sum())
print("Missing values in test:", test.isnull().sum().sum())
```

#### Save Prepared Data

Saving the cleaned training and test datasets for use in model training.

```python
# Save the cleaned datasets so they can be loaded directly in the modeling notebook

train.to_csv("../prepared_data/train_prepared.csv", index=False)
test.to_csv("../prepared_data/test_prepared.csv", index=False)

print("Prepared datasets saved successfully.")
print("Training:", train.shape)
print("Test:", test.shape)
```
